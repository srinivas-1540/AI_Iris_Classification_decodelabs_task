11.Inheritance
Inheritance allows one class to acquie properties and methods from another class.

Use inheritance when there is an IS-A relationship.

* Dog IS-A Animal
* Car IS-A Vehicle
* Developer IS-A Employee
* MAnager IS-A Employee

syntax:

class parent:
   pass

class child(parent):
    pass

```python
class Employee:
    def work(self):
        print("Employee is working")
class Developer(Employee):
  def code(self):
      print("Developer is coding")
d = Developer()
d.work()
d.code()
```

Encapsulation:
Encapsulation means combining data and methods into a class and controlling how the data is accessed.

* protect data
* Control access
* Prevent accidental modification
* Hide implementation details

Common Python access conventions:

* Public:self.name --nomal attribute.
* Protected-Style:self._name--conventionally indicates internal/protected use;
* Python does not strictly enforce it.
* Private-style:self.__balance--double underscore triggers python

```python
class BankAccount:
  def __init__(self, balance) :
      self.__balance = balance
  def deposit(self, amount):
      self.__balance += amount
  def get__balance(self) :
    return self.__balance

account = BankAccount(10000)
account.deposit(500)
print(account.get__balance())
```
